```python
# Load and prep the metadata

def load_profile_metadata(json_path):
    with open(json_path, "r") as f:
        meta = json.load(f)

    df_meta = (
        pd.DataFrame.from_dict(meta["variables"], orient="index")
          .reset_index()
          .rename(columns={"index": "variable"})
    )

    # Keep only the most useful columns for a "website-like" table
    keep = ["variable", "label", "concept", "group", "predicateType"]
    keep = [c for c in keep if c in df_meta.columns]
    df_meta = df_meta[keep]

    return df_meta
```

```python
# Turn the API pull inot "variables in rows, geos in columns"

def profile_to_wide_table(
    df_api,
    df_meta,
    geo_cols=("NAME", "state", "public use microdata area"),
    geo_label_col="NAME",
    include_metadata_cols=("label", "concept", "group"),
    value_cast="numeric",  # "numeric" or "keep"
):
    """
    df_api: DataFrame returned from api_response_to_df()
    df_meta: DataFrame returned from load_profile_metadata()

    geo_cols: columns that identify geography in the API response.
              For PUMAs you often have: state, public use microdata area, NAME.
    geo_label_col: which geo column to display as the column header (often NAME).
    include_metadata_cols: metadata fields to include as left-most columns in output.
    value_cast: cast values to numeric when possible.
    """

    geo_cols = [c for c in geo_cols if c in df_api.columns]

    # Identify which columns are actual variables (exclude geo identifiers)
    var_cols = [c for c in df_api.columns if c not in geo_cols]

    # Create a single geography key/label for the pivot columns
    if geo_label_col in df_api.columns:
        df_api = df_api.copy()
        df_api["geo"] = df_api[geo_label_col].astype(str)
    else:
        # fallback: concatenate geo parts
        df_api = df_api.copy()
        df_api["geo"] = df_api[geo_cols].astype(str).agg(" | ".join, axis=1)

    # Long form: (geo, variable, value)
    df_long = df_api.melt(
        id_vars=["geo"],
        value_vars=var_cols,
        var_name="variable",
        value_name="value"
    )

    # Optionally cast numeric
    if value_cast == "numeric":
        df_long["value"] = pd.to_numeric(df_long["value"], errors="ignore")
        # If you prefer strictly numeric with NaN for non-numeric:
        # df_long["value"] = pd.to_numeric(df_long["value"], errors="coerce")

    # Join metadata
    df_long = df_long.merge(df_meta, on="variable", how="left")

    # Build a row index: variable + chosen metadata columns
    row_cols = ["variable"] + [c for c in include_metadata_cols if c in df_long.columns]

    # Pivot: rows = variables, columns = geos
    wide = (
        df_long.pivot_table(
            index=row_cols,
            columns="geo",
            values="value",
            aggfunc="first"   # each (geo, variable) should be unique
        )
        .reset_index()
    )

    # Optional: sort columns nicely (metadata first, then geos)
    meta_part = row_cols
    geo_part = [c for c in wide.columns if c not in meta_part]
    wide = wide[meta_part + geo_part]

    return wide
```

```python
# Export to excel (or csv)
def export_tables(wide_df, out_xlsx="acs_profile_wide.xlsx", out_csv=None):
    # Excel
    with pd.ExcelWriter(out_xlsx, engine="openpyxl") as writer:
        wide_df.to_excel(writer, sheet_name="Profile (wide)", index=False)

    # Optional CSV (note: very wide CSVs can be awkward)
    if out_csv:
        wide_df.to_csv(out_csv, index=False)
```
