```python
# Clean dataframe
var_df = all_data.loc[:, ~all_data.columns.str.endswith(('EA', 'MA', 'PEA'))].copy()                   # drop annotation columns
var_df = var_df.replace(-888888888, np.nan)
var_df = var_df.replace(-999999999, np.nan)
var_df
```

```python
# labels
with open(r"C:\Users\HerreraMartinezR\OneDrive - NYC Office of Management and Budget\HTM Projects\ACS_Warehouse\docs\dp_variables_2022.json", 'r') as f:
    data = json.load(f)

labels2022 = (
    pd.DataFrame.from_dict(data["variables"], orient="index")
      .reset_index()
      .rename(columns={"index": "variable"})
)

with open(r"C:\Users\HerreraMartinezR\OneDrive - NYC Office of Management and Budget\HTM Projects\ACS_Warehouse\docs\dp_variables_2023.json", 'r') as f:
    data = json.load(f)

labels2023 = (
    pd.DataFrame.from_dict(data["variables"], orient="index")
      .reset_index()
      .rename(columns={"index": "variable"})
)

with open(r"C:\Users\HerreraMartinezR\OneDrive - NYC Office of Management and Budget\HTM Projects\ACS_Warehouse\docs\dp_variables_2024.json", 'r') as f:
    data = json.load(f)

labels2024 = (
    pd.DataFrame.from_dict(data["variables"], orient="index")
      .reset_index()
      .rename(columns={"index": "variable"})
)
```

```python
# Keep only DP02
labels2022 = labels2022[labels2022['group'] == 'DP02'].copy()
labels2023 = labels2023[labels2023['group'] == 'DP02'].copy()
labels2024 = labels2024[labels2024['group'] == 'DP02'].copy()
labels2022
```

```python
meta_labels_df = labels2024[['variable', 'label', 'group']]
meta_labels_df = meta_labels_df.rename(columns = {'variable' : 'var_name'})                                     # Change column names
meta_labels_df['var_type'] = meta_labels_df['var_name'].str.extract(r'\d+([A-Z]+)', expand=False)               # Gen var type
meta_labels_df['var_type'] = meta_labels_df['var_type'].replace({'E': 'Estimate','PE': 'Percentage'})           # Change type description
meta_labels_df['var_label'] = meta_labels_df['label'].replace('Estimate|Percent', '', regex=True)               # Gen base label
meta_labels_df = meta_labels_df.drop_duplicates()                                                               # Drop possible duplicates
meta_labels_df['var_label'] = meta_labels_df['var_label'].str.lstrip('!!')                                      # Remove initial '!!'
meta_labels_df['var_topic'] = meta_labels_df['var_label'].str.split('!!').str[0]                                # Add var topic
meta_labels_df['var_label'] = meta_labels_df['var_label'].str.replace('!!', ' - ', regex=True)                  # Remove other '!!'
meta_labels_df['var_attributes'] = meta_labels_df['var_name'].str[:-1] + 'M (marging of error)'                 # Gen column with var attributes
meta_labels_df = meta_labels_df[['var_name', 'var_type', 'var_topic', 'var_label', 'var_attributes', 'group']]
meta_labels_df
```

```python
# Save data
data_sample = 'acs1'
data_geography = 'PUMA'
data_group = 'DP02'
group_desc = 'SOCIAL_CHAR'


directory = fr'C:\Users\HerreraMartinezR\OneDrive - NYC Office of Management and Budget\Data\ACS\{data_sample}\panel\{data_geography}'
file_name = f'{data_sample}_paneL_2024_{data_geography}_NYC_{data_group}_{group_desc}.xlsx'

# Ensure the folder exists (create if missing)
os.makedirs(directory, exist_ok=True)


full_path = os.path.join(directory, file_name)

with pd.ExcelWriter(full_path) as writer:
    var_df.to_excel(writer, sheet_name = 'data', index = False)
    meta_labels_df.to_excel(writer, sheet_name = 'geo_labels', index = False)
```
