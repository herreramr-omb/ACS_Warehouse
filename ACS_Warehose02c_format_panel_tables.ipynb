```python
# Format labels
meta_labels_df = labels2024[['variable', 'label', 'group']]
meta_labels_df = meta_labels_df.rename(columns = {'variable' : 'var_name'})                                     # Change column names
meta_labels_df['var_type'] = meta_labels_df['var_name'].str.extract(r'\d+([A-Z]+)', expand=False)               # Gen var type
meta_labels_df['var_type'] = meta_labels_df['var_type'].replace({'E': 'Estimate','PE': 'Percentage'})           # Change type description
meta_labels_df['var_label'] = meta_labels_df['label'].replace('Estimate|Percent', '', regex=True)               # Gen base label
meta_labels_df = meta_labels_df.drop_duplicates()                                                               # Drop possible duplicates
meta_labels_df['var_label'] = meta_labels_df['var_label'].str.lstrip('!!')                                      # Remove initial '!!'
meta_labels_df['var_topic'] = meta_labels_df['var_label'].str.split('!!').str[0]                                # Add var topic
meta_labels_df['var_label'] = meta_labels_df['var_label'].str.replace('!!', ' - ', regex=True)                  # Remove other '!!'
meta_labels_df['var_attributes'] = meta_labels_df['var_name'].str[:-1] + 'M (marging of error)'                 # Gen column with var attributes
meta_labels_df = meta_labels_df[['var_name', 'var_type', 'var_topic', 'var_label', 'var_attributes', 'group']]  # Re-order columns
meta_labels_df = meta_labels_df.sort_values(by = 'var_name')                                                    # Sort values by var_name
meta_labels_df
```
